```python
import h5py
import numpy
import mediapy as media

# set path to your .h5 file and call inspect_h5(filename)
def inspect_h5(filename):
    """Open an HDF5 file and print its groups, datasets and attributes."""
    def walk_group(group, path=""):
        for key in group:
            obj = group[key]
            obj_path = f"{path}/{key}" if path else key
            if isinstance(obj, h5py.Group):
                print(f"{obj_path} (Group)")
                for a_key, a_val in obj.attrs.items():
                    print(f"  @{a_key}: {a_val}")
                walk_group(obj, obj_path)
            else:
                print(f"{obj_path} (Dataset) shape={obj.shape} dtype={obj.dtype}")
                for a_key, a_val in obj.attrs.items():
                    print(f"  @{a_key}: {a_val}")

    with h5py.File(filename, "r") as f:
        print(f"Opened: {filename}")
        print("Top-level keys:", list(f.keys()))
        walk_group(f)

inspect_h5("/home/sayantan/repos/tub/proj/explore/configs/fingerRamp_onRamp.h5")
```
